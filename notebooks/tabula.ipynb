```python
import tabula

import numpy as np
import pandas as pd

import re

import os
```

```python
# folder = "../data/barclays_statements/"
# file = "2024-04-05 Statement 05-APR-24 AC 43021297 08152120.pdf"
# file_path = folder + file
```

```python
folder = "../data/barclays_statements/"
for file in os.listdir(folder):
    file_path = folder + file

    dfs = tabula.read_pdf(file_path, pages='all')
    year = file.split("-")[0]
    year

    df = dfs[0].copy()
    df = df.drop("Unnamed: 0", axis=1)
    df.columns = ["Date Description", "Money out", "Money in", "Balance"]
    df = df.drop([0, 1], axis=0)

    df_clean = pd.DataFrame(columns = df.columns)
    for id, row in df.iterrows():
        if (row["Money out"] is np.nan) and (row["Money in"] is np.nan) and (row["Balance"] is np.nan):
            df_clean.loc[len(df_clean)-1, "Date Description"] += row["Date Description"]
        else:
            # df_clean = pd.concat([df_clean, row], axis=0)
            df_clean.loc[len(df_clean)] = row

    month_pattern=r"(^\d\d(?:(?:Jan|Feb)(?:uary)?|Mar(?:ch)?|Apr(?:il)?|May|Jun(?:e)?|Jul(?:y)?|Aug(?:ust)?|Sep(?:tember)?|Oct(?:ober)?|Nov(?:ember)?|Dec(?:ember)?))"

    df_clean["Date"] = None
    for id, row in df_clean.iterrows():
        split = re.search(month_pattern, row["Date Description"].replace(" ", ""))
        if split:
            df_clean.loc[id, "Date"] = split.group(0)+year
        else:
            df_clean.loc[id, "Date"] = df_clean.loc[id-1, "Date"]

    df_clean.to_csv("outputs/"+file.replace(".pdf", ".csv"))
```
